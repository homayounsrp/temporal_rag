```python
from graphiti_core import Graphiti
from config import NEO4J_URI, NEO4J_USER, NEO4J_PASSWORD, LLM_CLIENT
from Agent import run_query
```

## Initialize Graphiti

```python
graphiti = Graphiti(NEO4J_URI, NEO4J_USER, NEO4J_PASSWORD, LLM_CLIENT)

try:
    await graphiti.build_indices_and_constraints()
    print("✓ Graphiti initialized")
except:
    print("✓ Using existing indices")
```

## Query 1: Retrieve Only

Agent uses **search_graphiti** tool to find information.

```python
query1 = "What is the current role of Dr. Sarah Chen?"
print(f"Query: {query1}\n")

result1 = await run_query(graphiti, query1)
print(f"Response: {result1}")
```

## Query 2: Retrieve & Update

Agent automatically uses **both tools**:
1. `search_graphiti` - to find current information
2. `add_episode_to_graphiti` - to add new episode

```python
query2 = """Search for Dr. Sarah Chen's role, then add this episode: 
'In 18th of January 2026, Dr. Sarah Chen was promoted to Chief Scientist. and she is not Lead Marine Biologist anymre'"""
print(f"Query: {query2}\n")

result2 = await run_query(graphiti, query2)
print(f"Response: {result2}")
```

## Cleanup

```python
await graphiti.close()
print("✓ Connection closed")
```
